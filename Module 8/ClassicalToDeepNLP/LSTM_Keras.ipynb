```python
# Import libraries
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences


texts = [
    "I love this movie",
    "This film is amazing",
    "Very good acting",
    "Excellent story",
    "I hate this movie",
    "Terrible film",
    "Very boring story",
    "Worst acting ever"
]

# Labels
# 1 = Positive
# 0 = Negative

labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])


#words to numbers
tokenizer = Tokenizer()

# Learn vocabulary
tokenizer.fit_on_texts(texts)

# Convert sentences to sequences
sequences = tokenizer.texts_to_sequences(texts)

print("Word Index:")
print(tokenizer.word_index)

print("\nSequences:")
print(sequences)


# Make all sequences same length


max_length = 4

X = pad_sequences(sequences, maxlen=max_length)

print("\nPadded Sequences:")
print(X)


# Build LSTM model


model = Sequential()

# Embedding layer
# Converts word numbers into dense vectors

model.add(
    Embedding(
        input_dim=len(tokenizer.word_index) + 1,
        output_dim=8,
        input_length=max_length
    )
)

# LSTM layer
# Learns sequence patterns

model.add(LSTM(8))

# Output layer
# Sigmoid gives value between 0 and 1

model.add(Dense(1, activation='sigmoid'))


# Compile model


model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)


# Train model


model.fit(
    X,
    labels,
    epochs=50,
    verbose=1
)

# Test on new sentence


test_text = ["This movie is excellent"]

# Convert to sequence
test_seq = tokenizer.texts_to_sequences(test_text)

# Pad sequence
test_pad = pad_sequences(test_seq, maxlen=max_length)

# Predict
prediction = model.predict(test_pad)

print("\nPrediction:", prediction)

# Convert probability to label
if prediction > 0.5:
    print("Positive Review")
else:
    print("Negative Review")
```
