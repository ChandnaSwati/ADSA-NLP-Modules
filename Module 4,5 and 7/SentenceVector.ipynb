```python
# import necessary libraries
import gensim.downloader as api
import numpy as np

# Load pre-trained GloVe embeddings
glove = api.load("glove-wiki-gigaword-100")
```

```python
# Define a sample sentence
sentence = "machine learning improves business decisions"

# Preprocess the sentence: convert to lowercase and split into words
words = sentence.lower().split()


# Retrieve the vectors for each word in the sentence
vectors = []
for word in words:
    vector = glove[word]
    vectors.append(vector)
    print(f"{word:12} -> {vector[:5]}...")  # Print the first 5 dimensions of the vector


```
