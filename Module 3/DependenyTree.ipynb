```python
import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "Researchers analyze large datasets efficiently."

doc = nlp(text)

# Print dependency information
for token in doc:
    print(
        f"{token.text:<15}"
        f"{token.dep_:<10}"
        f"{token.head.text}"
    )

```

```python

from spacy import displacy
from IPython.display import HTML, display


# style="dep" tells displaCy to create a dependency parse visualization
# jupyter=False prevents spaCy from trying to automatically display
html = displacy.render(doc, style="dep", jupyter=False)


# Convert the HTML string into an HTML object and display it inside the Jupyter Notebook
display(HTML(html))
```
