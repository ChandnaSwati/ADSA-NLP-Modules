### FindingPatterns
We want to extract all numbers from the text.

\d+
- `\d` → matches any digit (0–9)
- `+` → means one or more occurrences
So:
`\d+`
matches:
- 45
- 12

```python
import re
text = "There are 45 students and 12 laptops."
numbers = re.findall(r'\d+', text)
print(numbers)
```

### Extract Email Addresses 

```python
text = """

Contact us at info@university.edu

or support@gmail.com

"""
## \S+ matches any non-whitespace characters before and after the @ symbol
emails = re.findall(r'\S+@\S+', text)

print(emails)
```

### Remove Punctuation

```python
text = "Hello!!! Welcome to NLP, class #2026."

clean_text = re.sub(r'[^\w\s]', '', text)

print(clean_text)
```

### Remove URLs from Social Media Text 

```python
text = "Check my website https://example.com now!"
## sub: find matching patterns and replace them.
clean = re.sub(r'http\S+', '', text)

print(clean)
```
