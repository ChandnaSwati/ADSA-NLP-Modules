```python
# Import different stemming algorithms from NLTK
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import LancasterStemmer

# Create stemmer objects
porter = PorterStemmer()
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()

# List of words to test
words = [
    "running",
    "studies",
    "happily",
    "university",
    "relational",
    "relationally",
    "happiness"
]

# Print table headers
# ljust(15) aligns the text into columns
print("WORD".ljust(15),
      "PORTER".ljust(15),
      "SNOWBALL".ljust(15),
      "LANCASTER")

# Print a separator line
print("-" * 60)

# Loop through each word in the list
for word in words:

    # Apply Porter stemming
    porter_stem = porter.stem(word)

    # Apply Snowball stemming
    snowball_stem = snowball.stem(word)

    # Apply Lancaster stemming
    lancaster_stem = lancaster.stem(word)

    # Print results in table format
    print(
        word.ljust(15),          # Original word
        porter_stem.ljust(15),  # Porter output
        snowball_stem.ljust(15),# Snowball output
        lancaster_stem          # Lancaster output
    )
```
