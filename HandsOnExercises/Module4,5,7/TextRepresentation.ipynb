# Business Scenario

A multinational company receives thousands of internship and job applications every year.

The company wants to develop an NLP-based talent matching system that automatically recommends the most suitable positions for applicants.

However, management is unsure whether the system should use:

- TF-IDF
- Word2Vec
- GloVe
- FastText

Your consulting team has been hired to investigate and provide a recommendation.

### Important

You are **NOT** asked to build the perfect recruitment system.

Your objective is to determine:

> Which embedding approach should be used and under what conditions?

You must support your recommendation using experiments, analysis, and critical evaluation.
