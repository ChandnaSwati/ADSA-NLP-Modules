```python
from gensim.models import Word2Vec, FastText

sentences = [
    ["machine", "learning", "improves", "business", "decisions"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["data", "science", "uses", "statistics", "and", "machine", "learning"],
    ["artificial", "intelligence", "supports", "automation"],
    ["data", "engineers", "build", "pipelines"],
    ["machine", "learning", "engineers", "build", "models"],
]

word2vec_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,
    epochs=100
)

fasttext_model = FastText(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,
    epochs=100
)

print("Models trained.")
```

```python
import gensim.downloader as api

# Load 100-dimensional GloVe embeddings
glove = api.load("glove-wiki-gigaword-100")

print("Model loaded successfully!")
```

```python
pairs = [
    ("doctor", "hospital"),
    ("doctor", "football"),
    ("king", "queen"),
    ("data", "science"),
    ("teacher", "university")
]

for w1, w2 in pairs:
    score = glove.similarity(w1, w2)
    print(f"{w1:10} -->  {w2:10}: {score:.3f}")
```

Task 2

1. Rank the word pairs from most similar to least similar.
2. Which pair receives the highest score?
3. Which pair receives the lowest score?
4. Does the ranking match your intuition?

```python
result = glove.most_similar(
    positive=["king", "woman"],
    negative=["man"],
    topn=1
)

print(result)
```

```python
analogy_tests = [
    (["paris", "germany"], ["france"]),
    (["teacher", "university"], ["school"]),
    (["doctor", "woman"], ["man"])
]

for positive, negative in analogy_tests:
    print("\nPositive:", positive)
    print("Negative:", negative)
    print(glove.most_similar(
        positive=positive,
        negative=negative,
        topn=2
    ))
```
