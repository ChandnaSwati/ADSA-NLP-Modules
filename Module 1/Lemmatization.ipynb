```python
# Import NLTK library
import nltk

# Import WordNet Lemmatizer
from nltk.stem import WordNetLemmatizer

# Download WordNet dictionary
nltk.download("wordnet")

# Download additional multilingual WordNet data
nltk.download("omw-1.4")

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()

# List of words for testing
words = [
    "running",
    "studies",
    "better",
    "wolves",
    "children"
]

# Loop through each word in the list
for word in words:

    # Apply lemmatization
    lemma = lemmatizer.lemmatize(word)

    # Print original word and lemmatized word
    print(word, "→", lemma)
```

```python
# Import spaCy library
import spacy

# Load the small English NLP model
# contains the necessary components for tokenization and lemmatization
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = "The sky is clear and stars are twinkling."

# Process the text using spaCy pipeline
doc = nlp(text)


# Print table header
print("TOKEN".ljust(15),
      "LEMMA".ljust(15),
      "POS TAG")

# Print separator line
print("-" * 40)

# Loop through each token (word) in the document
for token in doc:

    print(
       token.text.ljust(15),     # Original word
        token.lemma_.ljust(15),   # Lemmatized word
        token.pos_                # Part of Speech
    )
```
